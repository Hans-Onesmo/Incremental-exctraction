```python
from pathlib import Path

# Ensure directory exists
Path("loaded_data").mkdir(exist_ok=True)
```

```python
# Section 1: Load Setup
import pandas as pd

# Define file paths
full_csv_path = 'transformed_full.csv'
incremental_csv_path = 'transformed_incremental.csv'
full_parquet_path = 'loaded_data/full_data.parquet'
incremental_parquet_path = 'loaded_data/incremental_data.parquet'
```

```python
# Section 2: Load Full Transformed Data
full_df = pd.read_csv(full_csv_path)
full_df.to_parquet(full_parquet_path, index=False)
print("Full dataset saved as:", full_parquet_path)
```

```python
# Section 3: Load Incremental Transformed Data
incremental_df = pd.read_csv(incremental_csv_path)
incremental_df.to_parquet(incremental_parquet_path, index=False)
print("Incremental dataset saved as:", incremental_parquet_path)
```

```python
# Section 4: Verification
print("Preview of full_data.parquet:")
display(pd.read_parquet(full_parquet_path).head())

print("Preview of incremental_data.parquet:")
display(pd.read_parquet(incremental_parquet_path).head())
```
